```python
#import modules
import pandas as pd
import sys
from pathlib import Path # Ensure Path is imported if not already in config
from datetime import datetime

# ... (your existing sys.path and config imports)
sys.path = [
  "/home/abhishek/Documents/Study/dmml_assignment/group_4_dmml",
]
from config import USER_INTERACTION_DIR
```

```python
# read interactions data files 

year = datetime.now().strftime("%Y")
month = datetime.now().strftime("%m")
# day = datetime.now().strftime("%d")
day=10

# 1. Define the directory path (without the filename)
search_dir = USER_INTERACTION_DIR / f"year={year}" / f"month={month}" / f"day={day}"

# 2. Use glob to find all files matching the pattern
# This returns a generator of Path objects
files = list(search_dir.glob("interactions_*.csv"))

if not files:
    print(f"No files found in {search_dir}")
else:
    # 3. Read and concatenate all matched files
    df_list = [pd.read_csv(f) for f in files]
    df = pd.concat(df_list, ignore_index=True)
    
    print(f"Successfully loaded {len(files)} file(s).")
```

1. Data Cleaning & Preprocessing

```python
missing_pk = df[df['interaction_id'].isna()]
print(f"Total rows missing PK: {len(missing_pk)}")

total_rows = len(df)
missing_count = len(missing_pk)
percent_missing = (missing_count / total_rows) * 100

print(f"Percentage of missing PKs: {percent_missing:.2f}%")
```

```python
#Decision: Dropping records where keys are null
df_cleaned = df.dropna(subset=['interaction_id']).copy()
df_cleaned = df_cleaned.dropna(subset=['user_id']).copy()
df_cleaned = df_cleaned.dropna(subset=['product_id']).copy()

# 2. Reset the index (optional but recommended after dropping rows)
df_cleaned = df_cleaned.reset_index(drop=True)

df_cleaned.count()
```

```python
df_cleaned_missing_pk = df_cleaned[df_cleaned['user_id'].isna()]
print(f"Total rows missing PK after cleaning: {len(df_cleaned_missing_pk)}")
```
