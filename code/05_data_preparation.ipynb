*   Display the first 5 rows and column information for each to verify successful loaded and inspect initial data types.

```python
%matplotlib inline
#import modules
import pandas as pd
import sys
from pathlib import Path # Ensure Path is imported if not already in config
from datetime import datetime
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler

# ... (your existing sys.path and config imports)
sys.path = [
  "/home/abhishek/Documents/Study/dmml_assignment/group_4_dmml",
]
from config import USER_INTERACTION_DIR, USERS_DIR, PRODUCT_DIR,  PROCESSED_INTERACTION_DIR, \
    PROCESSED_USERS_DIR, PROCESSED_PRODUCT_DIR
```

```python
# read interactions data files

year = datetime.now().strftime("%Y")
month = datetime.now().strftime("%m")
day = datetime.now().strftime("%d")
# day=10

# 1. Define the directory path (without the filename)
interaction_search_dir = USER_INTERACTION_DIR / f"year={year}" / f"month={month}" / f"day={day}"
users_search_dir = USERS_DIR / f"year={year}" / f"month={month}" / f"day={day}"
products_search_dir = PRODUCT_DIR / f"year={year}" / f"month={month}" / f"day={day}"

# 2. Use glob to find all files matching the pattern
# This returns a generator of Path objects

#2.1 Load users dataset
files = list(users_search_dir.glob("users*.csv"))
if not files:
    print(f"No files found in {users_search_dir}")
else:
    # Read and concatenate all matched files
    df_list = [pd.read_csv(f) for f in files]
    users_df = pd.concat(df_list, ignore_index=True)
    print(f"Successfully loaded {len(files)} file(s).")
    print("\n--- Users DataFrame ---")
    print("First 5 rows:")
    display(users_df.head())
    print("Column information:")
    users_df.info()

#2.2 Load products dataset
files = list(products_search_dir.glob("products*.csv"))
if not files:
    print(f"No files found in {products_search_dir}")
else:
    # Read and concatenate all matched files
    df_list = [pd.read_csv(f) for f in files]
    products_df = pd.concat(df_list, ignore_index=True)
    print(f"Successfully loaded {len(files)} file(s).")
    print("\n--- Products DataFrame ---")
    print("First 5 rows:")
    display(products_df.head())
    print("Column information:")
    products_df.info()

#2.3 Load interaction dataset
files = list(interaction_search_dir.glob("interactions*.csv"))
if not files:
    print(f"No files found in {interaction_search_dir}")
else:
    # Read and concatenate all matched files
    df_list = [pd.read_csv(f) for f in files]
    interactions_df = pd.concat(df_list, ignore_index=True)
    print(f"Successfully loaded {len(files)} file(s).")
    print("\n--- Interactions DataFrame ---")
    print("First 5 rows:")
    display(interactions_df.head())
    print("Column information:")
    interactions_df.info()
```

# 2.   Initial Data Cleaning


*   Perform general data cleaning across all three datasets, including:
*   Handling duplicates
*   Correcting inconsistent data types
*   Addressing basic format errors in columns like dates or emails

```python
print("### 2.1 Handling Duplicates ###")

# users_df
initial_rows_users = users_df.shape[0]
users_df.drop_duplicates(inplace=True)
duplicates_removed_users = initial_rows_users - users_df.shape[0]
print(f"Users DataFrame: {duplicates_removed_users} duplicate rows removed. New shape: {users_df.shape}")

# products_df
initial_rows_products = products_df.shape[0]
products_df.drop_duplicates(inplace=True)
duplicates_removed_products = initial_rows_products - products_df.shape[0]
print(f"Products DataFrame: {duplicates_removed_products} duplicate rows removed. New shape: {products_df.shape}")

# interactions_df
initial_rows_interactions = interactions_df.shape[0]
interactions_df.drop_duplicates(inplace=True)
duplicates_removed_interactions = initial_rows_interactions - interactions_df.shape[0]
print(f"Interactions DataFrame: {duplicates_removed_interactions} duplicate rows removed. New shape: {interactions_df.shape}")
```

```python
print("\n### 2.2 Cleaning users_df ###")

# Convert 'age' to numeric
users_df['age'] = pd.to_numeric(users_df['age'], errors='coerce')

# Convert 'signup_date' to datetime
users_df['signup_date'] = pd.to_datetime(users_df['signup_date'], errors='coerce')

# Standardize 'gender' column
users_df['gender'] = users_df['gender'].str.lower().str.strip()
# Mapping variations if necessary (example, if 'm' and 'male' exist, map to 'male')
# For now, just lowercasing and stripping whitespace as per instruction

# Clean 'email' and 'city' columns by stripping whitespace
users_df['email'] = users_df['email'].str.strip()
users_df['city'] = users_df['city'].str.strip()

print("--- Cleaned Users DataFrame ---")
print("First 5 rows:")
display(users_df.head())
print("Column information:")
users_df.info()
```

```python
print("\n### 2.3 Cleaning products_df ###")

# Standardize 'category' and 'brand' columns
products_df['category'] = products_df['category'].str.strip().str.title()
products_df['brand'] = products_df['brand'].str.strip().str.title()

print("--- Cleaned Products DataFrame ---")
print("First 5 rows:")
display(products_df.head())
print("Column information:")
products_df.info()
```

```python
print("\n### 2.3 Cleaning interactions_df ###")

# Convert 'timestamp' to datetime
interactions_df['timestamp'] = pd.to_datetime(interactions_df['timestamp'], errors='coerce')

# Convert 'session_duration_sec' to numeric
interactions_df['session_duration_sec'] = pd.to_numeric(interactions_df['session_duration_sec'], errors='coerce')

# Standardize 'action' and 'device' columns
interactions_df['action'] = interactions_df['action'].str.strip().str.lower()
interactions_df['device'] = interactions_df['device'].str.strip().str.lower()

print("--- Cleaned Interactions DataFrame ---")
print("First 5 rows:")
display(interactions_df.head())
print("Column information:")
interactions_df.info()
```

# 3.   Handle Missing Data


*   Address missing values in all datasets.
*   Handle missing user-item interactions in the 'interactions' dataset.
*   Apply appropriate imputation or removal strategies for other missing values in 'users' and 'products' datasets.

```python
print("### Handling Missing Values ###")

# 1. For interactions_df
print("\n--- Interactions DataFrame: Missing values before cleaning ---")
print(interactions_df.isnull().sum())

# Drop rows where any critical interaction column has missing values
initial_interactions_rows = interactions_df.shape[0]
critical_interaction_cols = ['interaction_id', 'user_id', 'product_id', 'action', 'timestamp', 'device', 'session_duration_sec']
interactions_df.dropna(subset=critical_interaction_cols, inplace=True)
print(f"Dropped {initial_interactions_rows - interactions_df.shape[0]} rows with missing critical interaction data from interactions_df.")

print("\n--- Interactions DataFrame: Missing values after cleaning ---")
print(interactions_df.isnull().sum())

# 2. For users_df
print("\n--- Users DataFrame: Missing values before cleaning ---")
print(users_df.isnull().sum())

# Drop rows where user_id is missing
initial_users_rows = users_df.shape[0]
users_df.dropna(subset=['user_id'], inplace=True)
print(f"Dropped {initial_users_rows - users_df.shape[0]} rows with missing user_id from users_df.")

# Impute missing gender with mode
users_df['gender'] = users_df['gender'].fillna(users_df['gender'].mode()[0])

# Impute missing age with median
users_df['age'] = users_df['age'].fillna(users_df['age'].median())

# Impute missing city with mode
users_df['city'] = users_df['city'].fillna(users_df['city'].mode()[0])

# Impute missing name and email with 'unknown'
users_df['name'] = users_df['name'].fillna('unknown')
users_df['email'] = users_df['email'].fillna('unknown')

# Impute missing signup_date with mode
users_df['signup_date'] = users_df['signup_date'].fillna(users_df['signup_date'].mode()[0])

print("\n--- Users DataFrame: Missing values after cleaning ---")
print(users_df.isnull().sum())

# 3. For products_df
print("\n--- Products DataFrame: Missing values before cleaning ---")
print(products_df.isnull().sum())

# Drop rows where product_id is missing
initial_products_rows = products_df.shape[0]
products_df.dropna(subset=['product_id'], inplace=True)
print(f"Dropped {initial_products_rows - products_df.shape[0]} rows with missing product_id from products_df.")

# Impute missing category and brand with mode
products_df['category'] = products_df['category'].fillna(products_df['category'].mode()[0])
products_df['brand'] = products_df['brand'].fillna(products_df['brand'].mode()[0])

# Impute missing price, rating, stock_level with median
products_df['price'] = products_df['price'].fillna(products_df['price'].median())
products_df['rating'] = products_df['rating'].fillna(products_df['rating'].median())
products_df['stock_level'] = products_df['stock_level'].fillna(products_df['stock_level'].median())

print("\n--- Products DataFrame: Missing values after cleaning ---")
print(products_df.isnull().sum())
```

# 4.   Encode Categorical Attributes


*   Apply suitable encoding techniques
*   Apply One-Hot Encoding and Label Encoding techniques
*   To categorical columns such as 'gender', 'city', 'category', 'brand', and 'action'.



```python
print("\n### Encoding Categorical Attributes ###")

# 1. Encode users_df
print("\n--- Encoding users_df ---")
# categorical_cols_users = ['gender', 'city']
categorical_cols_users = ['gender']
cols_to_encode_users = [col for col in categorical_cols_users if col in users_df.columns]

if cols_to_encode_users:
    users_df = pd.get_dummies(users_df, columns=cols_to_encode_users, drop_first=True)
else:
    print(f"Columns {categorical_cols_users} not found in users_df, skipping encoding for these columns.")

print("First 5 rows of users_df after encoding:")
display(users_df.head())
print("Column information of users_df after encoding:")
users_df.info()

# 2. Encode products_df
print("\n--- Encoding products_df ---")
#categorical_cols_products = ['category', 'brand']
categorical_cols_products = ['category']
cols_to_encode_products = [col for col in categorical_cols_products if col in products_df.columns]

if cols_to_encode_products:
    products_df = pd.get_dummies(products_df, columns=cols_to_encode_products, drop_first=True)
else:
    print(f"Columns {categorical_cols_products} not found in products_df, skipping encoding for these columns.")

print("First 5 rows of products_df after encoding:")
display(products_df.head())
print("Column information of products_df after encoding:")
products_df.info()

# 3. Encode interactions_df
print("\n--- Encoding interactions_df ---")
categorical_cols_interactions = ['action', 'device']
cols_to_encode_interactions = [col for col in categorical_cols_interactions if col in interactions_df.columns]

if cols_to_encode_interactions:
    interactions_df = pd.get_dummies(interactions_df, columns=cols_to_encode_interactions, drop_first=True)
else:
    print(f"Columns {categorical_cols_interactions} not found in interactions_df, skipping encoding for these columns.")

print("First 5 rows of interactions_df after encoding:")
display(interactions_df.head())
print("Column information of interactions_df after encoding:")
interactions_df.info()
```

# 5.   Normalize Numerical Variables


*   Normalize numerical features like 'age', 'price', 'rating', 'stock_level', 'timestamp', and 'session_duration_sec' to a common scale
*   Prepare them for analysis



```python
print("\n### Normalizing Numerical Variables ###")

# 1. Convert 'signup_date' and 'timestamp' to numerical Unix timestamps
# For users_df
users_df['signup_date_unix'] = users_df['signup_date'].astype('int64') // 10**9

# For interactions_df
interactions_df['timestamp_unix'] = interactions_df['timestamp'].astype('int64') // 10**9

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# 2. Normalize users_df
print("\n--- Normalizing users_df ---")
users_df[['age', 'signup_date_unix']] = scaler.fit_transform(users_df[['age', 'signup_date_unix']])
print("First 5 rows of users_df after normalization:")
display(users_df.head())
print("Column information of users_df after normalization:")
users_df.info()

# 3. Normalize products_df
print("\n--- Normalizing products_df ---")
products_df[['price', 'rating', 'stock_level']] = scaler.fit_transform(products_df[['price', 'rating', 'stock_level']])
print("First 5 rows of products_df after normalization:")
display(products_df.head())
print("Column information of products_df after normalization:")
products_df.info()

# 4. Normalize interactions_df
print("\n--- Normalizing interactions_df ---")
interactions_df[['session_duration_sec', 'timestamp_unix']] = scaler.fit_transform(interactions_df[['session_duration_sec', 'timestamp_unix']])

# 5. Rename timestamp columns in interactions_df
print("\n### Renaming columns in interactions_df ###")
interactions_df.rename(columns={'timestamp': 'event_timestamp', 'timestamp_unix': 'event_timestamp_unix'}, inplace=True)
print("Columns renamed successfully. Displaying first 5 rows and info:")
display(interactions_df.head())
interactions_df.info()

print("\nFirst 5 rows of interactions_df after normalization:")
display(interactions_df.head())
print("Column information of interactions_df after normalization:")
interactions_df.info()
```

# 6.   EDA (Exploratory Data Analysis) and Visualization


*   Conduct exploratory data analysis to:
*   Understand interaction distributions (e.g., by action, device, session duration),
*   Identify item popularity,
*   Analyze sparsity patterns in user-item interactions.
*   Generate summary plots such as **histograms** for **numerical distributions**, **bar charts** for **categorical distributions**, and potentially **heatmaps** for **correlations**

```python
print("### Exploratory Data Analysis and Visualization ###")

# Set plot style for better aesthetics
sns.set_style("whitegrid")

# 1. For interactions_df
print("\n--- Analyzing Interactions Data ---")

# a. Distribution of interaction action types
action_cols = [col for col in interactions_df.columns if col.startswith('action_')]
action_counts = interactions_df[action_cols].sum()
action_labels = [col.replace('action_', '').replace('_', ' ').title() for col in action_cols]

plt.figure(figsize=(10, 6))
sns.barplot(x=action_labels, y=action_counts.values, hue=action_labels, palette='viridis', legend=False)
plt.title('Distribution of Interaction Action Types')
plt.xlabel('Action Type')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close()

# b. Distribution of interaction device types
device_cols = [col for col in interactions_df.columns if col.startswith('device_')]
device_counts = interactions_df[device_cols].sum()
device_labels = [col.replace('device_', '').replace('_', ' ').title() for col in device_cols]

plt.figure(figsize=(10, 6))
sns.barplot(x=device_labels, y=device_counts.values, hue=device_labels, palette='viridis', legend=False)
plt.title('Distribution of Interaction Device Types')
plt.xlabel('Device Type')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close()

# c. Histogram for session_duration_sec
plt.figure(figsize=(10, 6))
sns.histplot(interactions_df['session_duration_sec'], bins=50, kde=True, color='skyblue')
plt.title('Distribution of Session Duration (seconds)')
plt.xlabel('Session Duration (normalized seconds)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
plt.close()

# d. Top 10 most interacted product_ids
top_products = interactions_df['product_id'].value_counts().head(10)

plt.figure(figsize=(12, 7))
sns.barplot(x=top_products.index, y=top_products.values, hue=top_products.index, palette='magma', legend=False)
plt.title('Top 10 Most Interacted Products')
plt.xlabel('Product ID')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close()

# 2. For users_df
print("\n--- Analyzing Users Data ---")

# a. Histogram for age distribution
plt.figure(figsize=(10, 6))
sns.histplot(users_df['age'], bins=30, kde=True, color='lightcoral')
plt.title('Distribution of User Age')
plt.xlabel('Age (normalized)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
plt.close()

# 3. For products_df
print("\n--- Analyzing Products Data ---")

# a. Histogram for price distribution
plt.figure(figsize=(10, 6))
sns.histplot(products_df['price'], bins=30, kde=True, color='lightgreen')
plt.title('Distribution of Product Prices')
plt.xlabel('Price (normalized)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
plt.close()

# b. Correlation heatmap for numerical columns
correlation_matrix = products_df[['price', 'rating', 'stock_level']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Heatmap of Product Numerical Attributes')
plt.tight_layout()
plt.show()
plt.close()

print("\n✅ All visualizations completed successfully!")
```
